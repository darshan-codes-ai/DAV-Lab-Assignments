## Step 0 - Setup

```python
import seaborn as sns
import pandas as pd
import matplotlib as plt
import numpy as np
df = pd.read_excel("earthquake_1995-2023.csv.xlsx", engine="openpyxl")
```

## Step 1 - Boxplot Using seaborn 

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
alert_order = ['green', 'yellow', 'orange', 'red']

plot_df = df[df['alert'].notna() & (df['alert'] != 'unknown')]

sns.boxplot(
    data=plot_df,
    x='alert',
    y='magnitude',
    order=alert_order
)
plt.title('Magnitude Distribution by Alert Level (Boxplot)')
plt.xlabel('Alert Level')
plt.ylabel('Magnitude')
plt.tight_layout()
plt.show()
```

## Color Coded Box Plot


```python
import matplotlib.pyplot as plt
import seaborn as sns

alert_order = ['green', 'yellow', 'orange', 'red']
palette = {
    'green': '#2ecc71',
    'yellow': '#f1c40f',
    'orange': '#f39c12',
    'red': '#e74c3c'
}

plot_df = df[df['alert'].notna() & (df['alert'] != 'unknown')]

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=plot_df,
    x='alert',
    y='magnitude',
    order=alert_order,
    palette=palette,
    width=0.6,
    linewidth=1.2,
    fliersize=3
)

plt.title('Magnitude Distribution by Alert Level (Colour-Coded Boxplot)')
plt.xlabel('Alert Level')
plt.ylabel('Magnitude')
plt.tight_layout()
plt.show()
```

## Step 2 Violinplot


```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
alert_order = ['green', 'yellow', 'orange', 'red']

plot_df = df[df['alert'].notna() & (df['alert'] != 'unknown')]

sns.violinplot(
    data=plot_df,
    x='alert',
    y='magnitude',
    order=alert_order
)
plt.title('Magnitude Distribution by Alert Level (Boxplot)')
plt.xlabel('Alert Level')
plt.ylabel('Magnitude')
plt.tight_layout()
plt.show()
```

## Colour Coded Violinplot

```python
import matplotlib.pyplot as plt
import seaborn as sns

alert_order = ['green', 'yellow', 'orange', 'red']
palette = {
    'green': '#2ecc71',
    'yellow': '#f1c40f',
    'orange': '#f39c12',
    'red': '#e74c3c'
}

plot_df = df[df['alert'].notna() & (df['alert'] != 'unknown')]

plt.figure(figsize=(10, 6))
sns.violinplot(
    data=plot_df,
    x='alert',
    y='magnitude',
    order=alert_order,
    palette=palette,
    width=0.6,
    linewidth=1.2
)

plt.title('Magnitude Distribution by Alert Level (Colour-Coded Boxplot)')
plt.xlabel('Alert Level')
plt.ylabel('Magnitude')
plt.tight_layout()
plt.show()
```

## Step 3 - Histplot 


```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='magnitude', kde=True, bins=25, color='lightblue')
plt.title('Distribution of Earthquake Magnitudes (Histogram + KDE)')
plt.xlabel('Magnitude')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## Step 4 - HeatMap 


```python
nums_cols = ['magnitude', 'depth', 'cdi', 'mmi']
corr = df[nums_cols].corr()

plt.figure(figsize=(6.5, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f", square=True)
plt.title('Correlation Matrix (Magnitude, depth, cdi, mmi)')
plt.tight_layout()

plt.show()
```
